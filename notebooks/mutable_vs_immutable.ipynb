```python
numbers = [1, 2, 3]
history = [numbers]


def check(name, old):
    print(f"{name:<30} Same: {numbers is old}, ID: {id(numbers)}")
    history.append(numbers)


old = numbers
numbers.append(7)
check("append(7)", old)

old = numbers
numbers[0] = 10
check("numbers[0] = 10", old)

old = numbers
numbers[1:3] = [20, 30]
check("numbers[1:3] = [20, 30]", old)

old = numbers
numbers += [7, 8, 9]
check("numbers += [7, 8, 9]", old)

print()

old = numbers
numbers = numbers + [4, 5, 6]
check("numbers = numbers + [4, 5, 6]", old)

old = numbers
numbers = numbers[1:3]
check("numbers = numbers[1:3]", old)

old = numbers
numbers = sorted(numbers)
check("numbers = sorted(numbers)", old)
```

```python
a = [1, 2, 3]

# in place
def squares_of(numbers):
    for i, number in enumerate(numbers):
        numbers[i] = number**2
    return numbers

# new list
def squares_of2(numbers):
    result = []
    for number in numbers:
        result.append(number**2)
    return result

# list comprehension
def squares_of3(numbers):
    return [number**2 for number in numbers]

b = squares_of(a)
print(a is b)

b = squares_of2(a)
print(a is b)

b = squares_of3(a)
print(a is b)
```

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, major):
        super().__init__(name)
        self.major = major


john = Student("John", "Computer Science")
type(john)


print(type(john))
john.__class__ = Person  # type: ignore[assignment]
print(type(john))


print(john.name)
print(john.major) # major still there
print(isinstance(john, Student))  # False
print(isinstance(john, Person))   # True
```
